
# ➡Sales Data Analysis using NumPy, Pandas, Matplotlib

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


pd.set_option('display.max_columns', None)


data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West'],
    'Units_Sold': [120, 150, 170, 200, 160, 180, 140, 190, 210, 220, 250, 240],
    'Unit_Price': [50, 45, 55, 60, 52, 47, 50, 62, 58, 48, 56, 61],
    'Cost_Per_Unit': [30, 28, 35, 40, 32, 30, 33, 38, 36, 29, 34, 37]
}

df = pd.DataFrame(data)
print("Initial Data:\n", df)
print("\n--------------------------------------------")

# Step 3: Data Analysis using Pandas
df['Revenue'] = df['Units_Sold'] * df['Unit_Price']
df['Cost'] = df['Units_Sold'] * df['Cost_Per_Unit']
df['Profit'] = df['Revenue'] - df['Cost']

print("Data After Calculations:\n", df)
print("\n--------------------------------------------")

# Step 4: Basic Statistics using Pandas and NumPy
total_revenue = np.sum(df['Revenue'])
average_profit = np.mean(df['Profit'])
max_profit_month = df.loc[df['Profit'].idxmax(), 'Month']
min_profit_month = df.loc[df['Profit'].idxmin(), 'Month']

print(f"💰 Total Revenue: {total_revenue}")
print(f"📈 Average Monthly Profit: {average_profit:.2f}")
print(f"🏆 Highest Profit Month: {max_profit_month}")
print(f"📉 Lowest Profit Month: {min_profit_month}")
print("\n--------------------------------------------")

# Step 5: Regional Summary (Manual Aggregation)
region_summary = df.groupby('Region')[['Revenue', 'Profit']].sum().reset_index()
print("🌎 Regional Summary:\n", region_summary)
print("\n--------------------------------------------")

```

# ➡ Visualization using Matplotlib

```python


#  Monthly Revenue Line Chart
plt.figure(figsize=(10,5))
plt.plot(df['Month'], df['Revenue'], color='blue', marker='o')
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.grid(True)
plt.show()

#   Profit vs Month Bar Graph
plt.figure(figsize=(10,5))
plt.bar(df['Month'], df['Profit'], color='orange')
plt.title("Monthly Profit Analysis")
plt.xlabel("Month")
plt.ylabel("Profit ($)")
plt.show()

#  Region-wise Profit Comparison
plt.figure(figsize=(8,5))
plt.bar(region_summary['Region'], region_summary['Profit'], color='green')
plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit ($)")
plt.show()

#   Units Sold Distribution
plt.figure(figsize=(10,5))
plt.hist(df['Units_Sold'], bins=5, color='purple', edgecolor='black')
plt.title("Distribution of Units Sold")
plt.xlabel("Units Sold")
plt.ylabel("Frequency")
plt.show()

#   Revenue vs Cost Scatter Plot
plt.figure(figsize=(8,5))
plt.scatter(df['Revenue'], df['Cost'], color='red')
plt.title("Revenue vs Cost")
plt.xlabel("Revenue ($)")
plt.ylabel("Cost ($)")
plt.show()
```
